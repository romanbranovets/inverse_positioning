# Classic MI Estimator

This notebook estimates localization information without training a model. The endpoint variable is represented by an adaptive nested HEALPix endpoint hypothesis set.

For a chosen HEALPix level with $N$ endpoint hypotheses, the endpoint entropy cap is

$$
H(e)=\log_2 N.
$$

For each sampled trajectory, the notebook evaluates the known synthetic magnetic-field likelihood over the current endpoint hypotheses, builds the posterior table $p(e_i \mid \mathrm{traj})$, and estimates

$$
I(e; \mathrm{traj}) = H(e) - H(e \mid \mathrm{traj}).
$$

The endpoint resolution is increased per step count until the estimate is stable, has a small standard error, and is not close to the current entropy cap. The final $H(e)$ can therefore differ by step count.


## 1. Setup

The shared magnetic field $B(u)$, Earth-radius geometry, and noise constants are imported from `notebooks/shared`. The endpoint hypotheses use HEALPix levels up to `MAX_HEALPIX_NSIDE`. The levels are nested geometrically, which was originally used to reuse endpoint likelihoods between resolutions. The current evaluator instead uses a streaming regime: each `nside` is evaluated in endpoint chunks without storing the full likelihood table, so cross-`nside` likelihood reuse is intentionally disabled for memory efficiency.

## 2. Tractable Probability Table

For a fixed observed trajectory

$$
t = \{(b_\tau, \Delta u_\tau)\}_{\tau=0}^{S},
$$

where $b_\tau$ is the observed magnetic vector and $\Delta u_\tau$ is the observed meter step, the notebook builds a finite table over endpoint candidates

$$
\mathcal E = \{e_1,\ldots,e_N\}.
$$

### Directly Tractable Quantities

Two quantities are directly tractable in this synthetic setup.

First, the endpoint prior is chosen by construction. The endpoint grid is uniform, so

$$
p(e_i)=\frac{1}{N}.
$$

Second, the trajectory likelihood under a candidate endpoint is tractable because we know the magnetic field function and the observation noise model.

For each candidate endpoint $e_i$, reconstruct the candidate path backward from the endpoint using the observed meter steps:

$$
\hat u_{i,S}=e_i,
$$

$$
\hat u_{i,\tau}
=\operatorname{Exp}_{\hat u_{i,\tau+1}}(-\Delta u_\tau),
\qquad \tau=S-1,\ldots,0.
$$

The predicted magnetic observation at time $\tau$ is

$$
\hat b_{i,\tau}=B(\hat u_{i,\tau}).
$$

The sampler uses Gaussian magnetic noise:

$$
b_\tau = B(u_\tau)+\epsilon_\tau,
\qquad
\epsilon_\tau\sim\mathcal N(0,\sigma_B^2 I_3).
$$

Therefore the exact likelihood is

$$
p(t\mid e_i)
=
\prod_{\tau=0}^{S}
\mathcal N\!\left(
 b_\tau;
 B(\hat u_{i,\tau}),
 \sigma_B^2 I_3
\right).
$$

Expanded:

$$
p(t\mid e_i)
=
(2\pi\sigma_B^2)^{-\frac{3(S+1)}{2}}
\prod_{\tau=0}^{S}
\exp\left(
-\frac{1}{2}
\left\|\frac{b_\tau-B(\hat u_{i,\tau})}{\sigma_B}\right\|^2
\right).
$$

The Gaussian normalization constant

$$
(2\pi\sigma_B^2)^{-\frac{3(S+1)}{2}}
$$

does not depend on $e_i$. It cancels in the posterior normalization and in the likelihood-ratio term used for Localization Bits. So the code stores only the endpoint-dependent log-likelihood

$$
\ell_i
=
-\frac{1}{2}
\sum_{\tau=0}^{S}
\left\|\frac{b_\tau-B(\hat u_{i,\tau})}{\sigma_B}\right\|^2.
$$

### Derived Quantities

From the tractable table entries $p(e_i)$ and $p(t\mid e_i)$, the marginal trajectory probability is

$$
p(t)=\sum_{j=1}^{N}p(e_j)p(t\mid e_j).
$$

With a uniform prior and the stored log-likelihoods,

$$
\log p(t)
=
\operatorname{logsumexp}_{j}(\ell_j-\log N)+C,
$$

where $C$ is the omitted Gaussian constant.

The posterior endpoint table is then

$$
p(e_i\mid t)
=
\frac{p(e_i)p(t\mid e_i)}{p(t)}.
$$

Because $p(e_i)=1/N$, this is equivalently

$$
p(e_i\mid t)
=
\frac{\exp(\ell_i)}{\sum_{j=1}^{N}\exp(\ell_j)}.
$$

### Code Mapping

- `reconstruct_candidate_paths(...)` computes $\hat u_{i,\tau}$.
- `candidate_log_likelihoods(...)` computes $\ell_i$.
- `posterior_summary(...)` derives $p(t)$, $p(e_i\mid t)$, $H(e\mid t)$, and `Localization Bits` from the log-likelihood table.
- `summarize_observed_trajectory(...)` is the shared helper used by both MI evaluation and example generation.

The core posterior implementation is:

```python
log_prior = -math.log(prior_count)
log_p_traj = torch.logsumexp(log_likelihoods + log_prior, dim=0)
log_ratio = log_likelihoods - log_p_traj
log_posterior = log_prior + log_ratio
posterior = torch.exp(log_posterior)
```


````python
@torch.no_grad()
def reconstruct_candidate_paths(candidate_endpoints, observed_tokens, step_count):
    u = candidate_endpoints
    positions_rev = [u]
    for t in range(step_count - 1, -1, -1):
        du = observed_tokens[t, 6:8].expand(candidate_endpoints.shape[0], 2)
        e1, e2 = tangent_basis(u)
        back_step = -du[:, 0:1] * e1 - du[:, 1:2] * e2
        u = exp_map_sphere(u, back_step)
        positions_rev.append(u)
    return torch.stack(list(reversed(positions_rev)), dim=1)


@torch.no_grad()
def sample_endpoint_conditioned_trajectory(endpoint_grid, step_count, batch_size=1):
    endpoint_indices = torch.randint(
        endpoint_grid.shape[0], (batch_size,), device=DEVICE, dtype=torch.long
    )
    endpoints = endpoint_grid[endpoint_indices]
    max_tokens = step_count + 1
    x = torch.zeros(batch_size, max_tokens, FEATURE_DIM, device=DEVICE, dtype=DTYPE)
    path = torch.zeros(batch_size, max_tokens, 3, device=DEVICE, dtype=DTYPE)

    u = endpoints
    path[:, step_count, :] = u
    heading_2d = torch.randn(batch_size, 2, device=DEVICE, dtype=DTYPE)
    heading_2d = heading_2d / heading_2d.norm(dim=-1, keepdim=True).clamp_min(1e-8)

    for t in range(step_count - 1, -1, -1):
        noisy_direction = heading_2d + WALK_DIRECTION_RANDOMNESS * torch.randn(
            batch_size, 2, device=DEVICE, dtype=DTYPE
        )
        du_2d_meters = (
            STEP_METERS
            * noisy_direction
            / noisy_direction.norm(dim=-1, keepdim=True).clamp_min(1e-8)
        )
        observed_du = du_2d_meters + EPS_U_STD * torch.randn(
            batch_size, 2, device=DEVICE, dtype=DTYPE
        )
        x[:, t, 6:8] = observed_du

        e1, e2 = tangent_basis(u)
        back_step = -observed_du[:, 0:1] * e1 - observed_du[:, 1:2] * e2
        previous_u = exp_map_sphere(u, back_step)
        path[:, t, :] = previous_u
        heading_2d = noisy_direction / noisy_direction.norm(
            dim=-1, keepdim=True
        ).clamp_min(1e-8)
        u = previous_u

    observed_B = B(path.reshape(-1, 3)).reshape(batch_size, max_tokens, 3)
    observed_B = observed_B + EPS_B_STD * torch.randn_like(observed_B)
    x[:, :, :3] = observed_B
    x[:, 0, 3:6] = 0.0
    x[:, 1:, 3:6] = observed_B[:, 1:, :] - observed_B[:, :-1, :]
    if step_count > 0:
        x[:, :, 8:10] = torch.cat(
            [
                torch.zeros(batch_size, 1, 2, device=DEVICE, dtype=DTYPE),
                torch.cumsum(x[:, :-1, 6:8], dim=1),
            ],
            dim=1,
        )
    token_time = torch.linspace(0.0, 1.0, max_tokens, device=DEVICE, dtype=DTYPE)
    time_phase = 2.0 * math.pi * token_time
    x[:, :, 10] = token_time
    x[:, :, 11] = torch.sin(time_phase)
    x[:, :, 12] = torch.cos(time_phase)
    x[:, -1, 13] = 1.0
    return x, endpoints, endpoint_indices, path


@torch.no_grad()
def candidate_log_likelihoods(
    candidate_endpoints, observed_tokens, step_count, show_progress=False
):
    observed_B = observed_tokens[: step_count + 1, :3].reshape(1, -1)
    chunks = []
    starts = range(0, candidate_endpoints.shape[0], CANDIDATE_CHUNK_SIZE)
    if show_progress:
        starts = tqdm(starts, desc=f"endpoint chunks, steps={step_count}", leave=False)
    for start in starts:
        stop = min(start + CANDIDATE_CHUNK_SIZE, candidate_endpoints.shape[0])
        candidates = candidate_endpoints[start:stop]
        paths = reconstruct_candidate_paths(candidates, observed_tokens, step_count)
        predicted_B = B(paths.reshape(-1, 3)).reshape(candidates.shape[0], -1)
        residual = (observed_B - predicted_B) / EPS_B_STD
        chunks.append(-0.5 * residual.square().sum(dim=1))
    return torch.cat(chunks, dim=0)


@torch.no_grad()
def posterior_summary(log_likelihoods, prior_count=None, return_posterior=True):
    if prior_count is None:
        prior_count = int(log_likelihoods.numel())
    h_bits = math.log2(prior_count)
    log_prior = -math.log(prior_count)
    log_p_traj = torch.logsumexp(log_likelihoods + log_prior, dim=0)
    log_ratio = log_likelihoods - log_p_traj
    log_posterior = log_prior + log_ratio
    posterior = torch.exp(log_posterior) if return_posterior else None
    if posterior is None:
        weights = torch.softmax(log_likelihoods.double(), dim=0).to(
            log_likelihoods.dtype
        )
        entropy_bits = float((-(weights * log_posterior).sum() / math.log(2.0)).cpu())
        localization_bits = float((weights * (log_ratio / math.log(2.0))).sum().cpu())
    else:
        entropy_bits = float((-(posterior * log_posterior).sum() / math.log(2.0)).cpu())
        localization_bits = float((posterior * (log_ratio / math.log(2.0))).sum().cpu())
    return {
        "posterior": posterior,
        "log_p_traj": log_p_traj,
        "h_bits": h_bits,
        "h_cond_bits": entropy_bits,
        "localization_bits": localization_bits,
    }


@torch.no_grad()
def summarize_observed_trajectory(
    endpoint_grid,
    observed_tokens,
    step_count,
    show_progress=False,
    return_posterior=True,
):
    log_likelihoods = candidate_log_likelihoods(
        endpoint_grid, observed_tokens, step_count, show_progress=show_progress
    )
    summary = posterior_summary(
        log_likelihoods,
        prior_count=endpoint_grid.shape[0],
        return_posterior=return_posterior,
    )
    summary["log_likelihoods"] = log_likelihoods
    return summary


@torch.no_grad()
def reconstruct_candidate_paths_batch(
    candidate_endpoints, observed_tokens_batch, step_count
):
    batch_size = observed_tokens_batch.shape[0]
    candidate_count = candidate_endpoints.shape[0]
    u = candidate_endpoints.unsqueeze(0).expand(batch_size, candidate_count, 3).clone()
    positions_rev = [u]
    for t in range(step_count - 1, -1, -1):
        du = (
            observed_tokens_batch[:, t, 6:8]
            .unsqueeze(1)
            .expand(batch_size, candidate_count, 2)
        )
        flat_u = u.reshape(-1, 3)
        e1, e2 = tangent_basis(flat_u)
        e1 = e1.reshape(batch_size, candidate_count, 3)
        e2 = e2.reshape(batch_size, candidate_count, 3)
        back_step = -du[:, :, 0:1] * e1 - du[:, :, 1:2] * e2
        u = exp_map_sphere(flat_u, back_step.reshape(-1, 3)).reshape(
            batch_size, candidate_count, 3
        )
        positions_rev.append(u)
    return torch.stack(list(reversed(positions_rev)), dim=2)


@torch.no_grad()
def candidate_log_likelihoods_batch(
    candidate_endpoints,
    observed_tokens_batch,
    step_count,
    show_progress=False,
):
    observed_tokens_batch = observed_tokens_batch.to(DEVICE)
    batch_size = observed_tokens_batch.shape[0]
    out = torch.empty(
        batch_size, candidate_endpoints.shape[0], dtype=DTYPE, device="cpu"
    )
    starts = range(0, candidate_endpoints.shape[0], CANDIDATE_CHUNK_SIZE)
    if show_progress:
        starts = tqdm(starts, desc=f"endpoint chunks, steps={step_count}", leave=False)
    for start in starts:
        stop = min(start + CANDIDATE_CHUNK_SIZE, candidate_endpoints.shape[0])
        candidates = candidate_endpoints[start:stop].to(DEVICE)
        for batch_start in range(0, batch_size, TRAJECTORY_BATCH_SIZE):
            batch_stop = min(batch_start + TRAJECTORY_BATCH_SIZE, batch_size)
            tokens = observed_tokens_batch[batch_start:batch_stop]
            paths = reconstruct_candidate_paths_batch(candidates, tokens, step_count)
            predicted_B = B(paths.reshape(-1, 3)).reshape(
                batch_stop - batch_start, candidates.shape[0], -1
            )
            observed_B = tokens[:, : step_count + 1, :3].reshape(
                batch_stop - batch_start, 1, -1
            )
            residual = (observed_B - predicted_B) / EPS_B_STD
            out[batch_start:batch_stop, start:stop] = (
                (-0.5 * residual.square().sum(dim=2)).detach().cpu()
            )
    return out


@torch.no_grad()
def posterior_summary_batch(log_likelihoods, prior_count=None):
    if prior_count is None:
        prior_count = int(log_likelihoods.shape[1])
    h_bits = math.log2(prior_count)
    log_likelihoods = log_likelihoods.to(DEVICE)
    log_prior = -math.log(prior_count)
    log_p_traj = torch.logsumexp(log_likelihoods + log_prior, dim=1, keepdim=True)
    log_ratio = log_likelihoods - log_p_traj
    log_posterior = log_prior + log_ratio
    weights = torch.softmax(log_likelihoods, dim=1)
    h_cond_bits = (-(weights * log_posterior).sum(dim=1) / math.log(2.0)).detach().cpu()
    localization_bits = (
        (weights * (log_ratio / math.log(2.0))).sum(dim=1).detach().cpu()
    )
    return {
        "h_bits": h_bits,
        "h_cond_bits": h_cond_bits.numpy(),
        "localization_bits": localization_bits.numpy(),
    }


@torch.no_grad()
def compute_level_log_likelihoods_batch(
    max_endpoint_grid,
    observed_tokens_batch,
    step_count,
    nside,
    previous_state=None,
):
    active_indices = nested_subset_indices(nside, MAX_HEALPIX_NSIDE)
    active_count = int(active_indices.numel())
    batch_size = observed_tokens_batch.shape[0]

    if previous_state is None:
        missing_positions = torch.arange(active_count, dtype=torch.long)
        log_likelihoods = torch.empty(
            batch_size, active_count, dtype=DTYPE, device="cpu"
        )
    else:
        previous_indices = previous_state["active_indices"]
        previous_positions = torch.searchsorted(active_indices, previous_indices)
        if not torch.equal(active_indices[previous_positions], previous_indices):
            raise RuntimeError(
                "previous HEALPix endpoint set is not nested in current set"
            )
        keep_mask = torch.zeros(active_count, dtype=torch.bool)
        keep_mask[previous_positions] = True
        missing_positions = torch.nonzero(~keep_mask, as_tuple=False).flatten()
        log_likelihoods = torch.empty(
            batch_size, active_count, dtype=DTYPE, device="cpu"
        )
        log_likelihoods[:, previous_positions] = previous_state["log_likelihoods"]

    if missing_positions.numel() > 0:
        missing_global_indices = active_indices[missing_positions].to(DEVICE)
        missing_endpoints = max_endpoint_grid[missing_global_indices]
        missing_ll = candidate_log_likelihoods_batch(
            missing_endpoints, observed_tokens_batch, step_count
        )
        log_likelihoods[:, missing_positions] = missing_ll

    return {
        "nside": int(nside),
        "active_indices": active_indices,
        "log_likelihoods": log_likelihoods,
    }


@torch.no_grad()
def summarize_mi_samples(mi_samples, h_cond_samples, nside):
    mi = np.asarray(mi_samples, dtype=np.float64)
    h_cond = np.asarray(h_cond_samples, dtype=np.float64)
    h_bits = healpix_entropy_bits(nside)
    return {
        "nside": int(nside),
        "endpoint_count": int(12 * nside * nside),
        "h_bits": float(h_bits),
        "mi_bits_mean": float(mi.mean()),
        "mi_bits_variance": float(mi.var(ddof=1)) if mi.size > 1 else 0.0,
        "mi_bits_std": float(mi.std(ddof=1)) if mi.size > 1 else 0.0,
        "mi_bits_se": (
            float(mi.std(ddof=1) / math.sqrt(mi.size)) if mi.size > 1 else 0.0
        ),
        "h_cond_bits_mean": float(h_cond.mean()),
        "samples": int(mi.size),
    }


@torch.no_grad()
def choose_adaptive_level(level_rows):
    for i, row in enumerate(level_rows):
        if i + 1 < ADAPTIVE_MIN_LEVELS:
            continue
        previous = level_rows[i - 1]
        stable = (
            abs(row["mi_bits_mean"] - previous["mi_bits_mean"])
            <= ADAPTIVE_STABILITY_BITS
        )
        ci_small = 1.96 * row["mi_bits_se"] <= ADAPTIVE_CI_HALF_WIDTH_BITS
        not_cap_limited = (
            row["mi_bits_mean"] <= row["h_bits"] - ADAPTIVE_CAP_MARGIN_BITS
        )
        if stable and ci_small and not_cap_limited:
            row = dict(row)
            row["stop_reason"] = "stable_ci"
            return row
    row = dict(level_rows[-1])
    row["stop_reason"] = "max_level"
    return row


@torch.no_grad()
def candidate_chunk_log_likelihoods_batch(
    candidate_endpoints, observed_tokens_batch, step_count
):
    tokens = observed_tokens_batch.to(DEVICE)
    candidates = candidate_endpoints.to(DEVICE)
    paths = reconstruct_candidate_paths_batch(candidates, tokens, step_count)
    predicted_B = B(paths.reshape(-1, 3)).reshape(
        tokens.shape[0], candidates.shape[0], -1
    )
    observed_B = tokens[:, : step_count + 1, :3].reshape(tokens.shape[0], 1, -1)
    residual = (observed_B - predicted_B) / EPS_B_STD
    return -0.5 * residual.square().sum(dim=2)


@torch.no_grad()
def streaming_mi_samples_for_nside(
    max_endpoint_grid, observed_tokens_batch, step_count, nside
):
    active_indices = nested_subset_indices(nside, MAX_HEALPIX_NSIDE)
    endpoint_count = int(active_indices.numel())
    batch_size = int(observed_tokens_batch.shape[0])

    running_max = torch.full((batch_size,), -torch.inf, dtype=torch.float64)
    running_sum = torch.zeros(batch_size, dtype=torch.float64)
    running_sum_ll = torch.zeros(batch_size, dtype=torch.float64)

    starts = range(0, endpoint_count, CANDIDATE_CHUNK_SIZE)
    for start in starts:
        stop = min(start + CANDIDATE_CHUNK_SIZE, endpoint_count)
        candidate_indices = active_indices[start:stop].to(DEVICE)
        candidates = max_endpoint_grid[candidate_indices]

        for batch_start in range(0, batch_size, TRAJECTORY_BATCH_SIZE):
            batch_stop = min(batch_start + TRAJECTORY_BATCH_SIZE, batch_size)
            ll = (
                candidate_chunk_log_likelihoods_batch(
                    candidates,
                    observed_tokens_batch[batch_start:batch_stop],
                    step_count,
                )
                .double()
                .cpu()
            )

            chunk_max = ll.max(dim=1).values
            old_max = running_max[batch_start:batch_stop]
            new_max = torch.maximum(old_max, chunk_max)
            old_scale = torch.exp(old_max - new_max)
            exp_ll = torch.exp(ll - new_max[:, None])

            running_sum[batch_start:batch_stop] = running_sum[
                batch_start:batch_stop
            ] * old_scale + exp_ll.sum(dim=1)
            running_sum_ll[batch_start:batch_stop] = running_sum_ll[
                batch_start:batch_stop
            ] * old_scale + (exp_ll * ll).sum(dim=1)
            running_max[batch_start:batch_stop] = new_max

    log_z = running_max + torch.log(running_sum)
    expected_ll = running_sum_ll / running_sum.clamp_min(
        torch.finfo(torch.float64).tiny
    )
    h_bits = math.log2(endpoint_count)
    mi_samples = (
        (expected_ll - log_z + math.log(endpoint_count)) / math.log(2.0)
    ).numpy()
    h_cond_samples = h_bits - mi_samples
    return mi_samples, h_cond_samples


@torch.no_grad()
def estimate_step_count_adaptive(max_endpoint_grid, step_count, trajectory_count):
    x, endpoints, endpoint_indices, path = sample_endpoint_conditioned_trajectory(
        max_endpoint_grid, step_count, batch_size=trajectory_count
    )
    level_rows = []

    for nside in HEALPIX_NSIDE_LEVELS:
        mi_samples, h_cond_samples = streaming_mi_samples_for_nside(
            max_endpoint_grid, x, step_count, nside
        )
        level_row = summarize_mi_samples(mi_samples, h_cond_samples, nside)
        level_rows.append(level_row)

        selected = choose_adaptive_level(level_rows)
        print(
            f"steps={step_count}, nside={nside}, "
            f"I={level_row['mi_bits_mean']:.2f} +/- {1.96 * level_row['mi_bits_se']:.2f} bits, "
            f"H(e)={level_row['h_bits']:.2f}"
        )
        if selected["nside"] == nside and selected.get("stop_reason") == "stable_ci":
            break

    selected = choose_adaptive_level(level_rows)
    selected.update(
        {
            "steps": int(step_count),
            "distance_m": float(step_count * STEP_METERS),
            "level_history": level_rows,
            "memory_mode": "streaming_logsumexp",
        }
    )
    return selected


@torch.no_grad()
def example_for_step_count(max_endpoint_grid, nside, step_count):
    endpoint_indices = nested_subset_indices(nside, MAX_HEALPIX_NSIDE)
    endpoint_grid = max_endpoint_grid[endpoint_indices.to(DEVICE)]
    x, endpoint, endpoint_index, path = sample_endpoint_conditioned_trajectory(
        endpoint_grid, step_count, batch_size=1
    )
    summary = summarize_observed_trajectory(
        endpoint_grid, x[0], step_count, show_progress=False
    )
    true_probability = float(summary["posterior"][endpoint_index[0]].detach().cpu())
    true_rank = int(
        (summary["posterior"] > summary["posterior"][endpoint_index[0]]).sum().item()
        + 1
    )
    return {
        "steps": int(step_count),
        "nside": int(nside),
        "x": x[0],
        "path": path[0].detach().cpu(),
        "endpoint": endpoint[0].detach().cpu(),
        "endpoint_index": int(endpoint_index[0].detach().cpu()),
        "endpoint_indices": endpoint_indices.detach().cpu(),
        "posterior": summary["posterior"].detach().cpu(),
        "true_probability": true_probability,
        "true_rank": true_rank,
        "localization_bits": summary["localization_bits"],
        "h_cond_bits": summary["h_cond_bits"],
    }
````

## 3. Endpoint Hypothesis Pool

The finest endpoint pool is a nested HEALPix grid with `MAX_HEALPIX_NSIDE`. Each adaptive level corresponds to a nested subset of this same pool, so the endpoint hypotheses are comparable across resolutions.

## 4. Mutual Information Evaluation

For a chosen endpoint hypothesis set with $N$ endpoints, the quantity we want is the mutual information between endpoint and trajectory:

$$
I(e;t)
=
\mathbb E_t\left[
D_{\mathrm{KL}}\bigl(p(e\mid t)\,\|\,p(e)\bigr)
\right].
$$

Equivalently,

$$
I(e;t)=H(e)-H(e\mid t).
$$

The endpoint entropy cap is no longer a single global constant. For a HEALPix level with `nside`,

$$
N=12\,\texttt{nside}^2,
\qquad
H(e)=\log_2 N.
$$

For one sampled trajectory $t_m$, the posterior table conceptually defines the conditional entropy

$$
H(e\mid t_m)
=
-\sum_{i=1}^{N}p(e_i\mid t_m)\log_2 p(e_i\mid t_m).
$$

The per-trajectory information sample at that level is

$$
I_m=H(e)-H(e\mid t_m).
$$

The current evaluator tries increasing HEALPix levels, but it does **not** reuse likelihood values from previous `nside` levels. Reuse was attempted earlier because the endpoint sets are nested, but the required coarse-level likelihood caches became too large at high `nside`. The current implementation recomputes each `nside` in a streaming way, trading extra compute for bounded memory.


### Adaptive Computation

For each step count, the estimator starts from the coarsest HEALPix endpoint hypothesis set and then tries finer `nside` levels. After each level it recomputes the Monte Carlo mean MI, its standard error, and the current entropy cap $H(e)=\log_2 N$.

The evaluation path is memory-streamed: it does not materialize the full `trajectory_count * endpoint_count` likelihood table. Instead, for each `nside`, it scans endpoint chunks and maintains streaming log-sum-exp accumulators per trajectory. This computes the same posterior entropy/MI moments needed for evaluation while keeping memory bounded by roughly `TRAJECTORY_BATCH_SIZE * CANDIDATE_CHUNK_SIZE * (steps + 1)` plus small per-trajectory accumulators.

This means there is currently no cross-`nside` likelihood reuse in the main evaluation. The notebook originally tried to exploit nested HEALPix levels for reuse, but that required retaining large likelihood tables and caused memory pressure at high resolutions. The streaming regime discards chunk likelihoods immediately after updating the accumulators, which is slower but much more memory efficient. The full posterior table is still built only in the example visualization path, where it is used for a single trajectory.

## 5. Example Trajectory Posterior Maps

For one trajectory $t$, the exact posterior is computed on the finite endpoint grid:

$$
p(e_i\mid t)
=
\frac{\exp(\ell_i)}{\sum_j\exp(\ell_j)}.
$$

The exact `Localization Bits` value is also computed on that full endpoint grid. For visualization, the notebook renders a continuous pole-free latitude/longitude sphere and colors it by smoothly splatting the exact endpoint-grid posterior mass onto the rendered surface. This keeps the plotted high-probability regions tied to the same posterior table used for the information calculation, while avoiding Plotly mesh artifacts and posterior-dot clutter.

The displayed **Localization Bits** for that trajectory are

$$
\operatorname{LB}(t)
=
D_{\mathrm{KL}}\bigl(p(e\mid t)\,\|\,p(e)\bigr).
$$

Expanding the KL:

$$
\operatorname{LB}(t)
=
\sum_{i=1}^{N}p(e_i\mid t)
\log_2\frac{p(e_i\mid t)}{p(e_i)}.
$$

Using Bayes' rule,

$$
\frac{p(e_i\mid t)}{p(e_i)}
=
\frac{p(t\mid e_i)}{p(t)},
$$

so the same table gives

$$
\operatorname{LB}(t)
=
\sum_{i=1}^{N}p(e_i\mid t)
\log_2\frac{p(t\mid e_i)}{p(t)}
=
H(e)-H(e\mid t).
$$

In code, `example_for_step_count(...)` computes the exact posterior table once. The plotting cell then calls `posterior_surface_color(...)` only to render a smooth continuous approximation of that discrete posterior on the sphere.

The connection to mutual information is direct:

$$
I(e;t)=\mathbb E_t[\operatorname{LB}(t)].
$$

So each example plot shows one realized posterior localization map, while the evaluation plot estimates the average information gain over many sampled trajectories.


````python
def sphere_surface(lat_count=SURFACE_LAT_COUNT, lon_count=SURFACE_LON_COUNT):
    theta = (np.arange(lat_count, dtype=np.float64) + 0.5) * (np.pi / lat_count)
    phi = np.linspace(0.0, 2.0 * np.pi, lon_count + 1, dtype=np.float64)
    theta_grid, phi_grid = np.meshgrid(theta, phi, indexing="ij")
    surface = np.stack(
        [
            np.sin(theta_grid) * np.cos(phi_grid),
            np.sin(theta_grid) * np.sin(phi_grid),
            np.cos(theta_grid),
        ],
        axis=-1,
    )
    return surface


def posterior_surface_color(
    posterior,
    endpoint_indices,
    endpoint_index,
    top_k=PLOT_POSTERIOR_TOP_K,
    bandwidth_deg=1.6,
    chunk_size=4096,
):
    posterior_np = np.asarray(posterior, dtype=np.float64)
    endpoint_indices_np = np.asarray(endpoint_indices, dtype=np.int64)
    top_k = min(top_k, posterior_np.size)
    top_indices = np.argpartition(posterior_np, -top_k)[-top_k:]
    top_indices = np.unique(np.concatenate([top_indices, np.asarray([endpoint_index])]))

    centers_np = max_endpoint_grid_cpu[endpoint_indices_np[top_indices]]
    centers = torch.as_tensor(centers_np, device=DEVICE, dtype=DTYPE)
    masses = torch.as_tensor(posterior_np[top_indices], device=DEVICE, dtype=DTYPE)
    surface_flat = torch.as_tensor(
        surface_xyz.reshape(-1, 3), device=DEVICE, dtype=DTYPE
    )

    sigma = math.radians(bandwidth_deg)
    kappa = 1.0 / max(sigma * sigma, 1e-12)
    colors = []
    for start in range(0, surface_flat.shape[0], chunk_size):
        points = surface_flat[start : start + chunk_size]
        cosine = (points @ centers.T).clamp(-1.0, 1.0)
        kernel = torch.exp(kappa * (cosine - 1.0))
        colors.append(kernel @ masses)
    color = torch.cat(colors).detach().cpu().numpy()
    return color.reshape(surface_xyz.shape[:2])


surface_xyz = sphere_surface()
surface_x = surface_xyz[..., 0]
surface_y = surface_xyz[..., 1]
surface_z = surface_xyz[..., 2]
selected_nside_by_steps = {row["steps"]: row["nside"] for row in results}

examples = []
for step_count in tqdm(EVAL_STEP_COUNTS, desc="example maps"):
    examples.append(
        example_for_step_count(
            max_endpoint_grid, selected_nside_by_steps[step_count], step_count
        )
    )

subplot_titles = [
    f"{item['steps']} steps<br><sup>Localization Bits: {item['localization_bits']:.2f}, nside={item['nside']}</sup>"
    for item in examples
]
fig = make_subplots(
    rows=2,
    cols=3,
    specs=[
        [{"type": "scene"}, {"type": "scene"}, {"type": "scene"}],
        [{"type": "scene"}, {"type": "scene"}, {"type": "scene"}],
    ],
    subplot_titles=subplot_titles,
    horizontal_spacing=0.02,
    vertical_spacing=0.08,
)

for idx, item in enumerate(examples):
    row = idx // 3 + 1
    col = idx % 3 + 1
    surface_color = posterior_surface_color(
        item["posterior"], item["endpoint_indices"], item["endpoint_index"]
    )
    positive = surface_color[surface_color > 0.0]
    cmax = float(np.quantile(positive, 0.997)) if positive.size else 1.0
    cmax = max(cmax, 1e-14)
    path_xyz = item["path"].numpy() * 1.035
    endpoint_xyz = item["endpoint"].numpy() * 1.06

    fig.add_trace(
        go.Surface(
            x=surface_x,
            y=surface_y,
            z=surface_z,
            surfacecolor=surface_color,
            colorscale="Turbo",
            cmin=0.0,
            cmax=cmax,
            opacity=1.0,
            showscale=idx == len(examples) - 1,
            colorbar=(
                dict(title="smoothed p(e | traj)", len=0.72)
                if idx == len(examples) - 1
                else None
            ),
            name=f"posterior surface {item['steps']} steps",
            hovertemplate="smoothed p(e | traj)=%{surfacecolor:.3e}<extra></extra>",
        ),
        row=row,
        col=col,
    )
    fig.add_trace(
        go.Scatter3d(
            x=path_xyz[:, 0],
            y=path_xyz[:, 1],
            z=path_xyz[:, 2],
            mode="lines+markers",
            line=dict(color="white", width=6),
            marker=dict(size=4, color="white", line=dict(color="black", width=1)),
            name=f"trajectory {item['steps']} steps",
            showlegend=False,
            hovertemplate="trajectory<extra></extra>",
        ),
        row=row,
        col=col,
    )
    fig.add_trace(
        go.Scatter3d(
            x=[endpoint_xyz[0]],
            y=[endpoint_xyz[1]],
            z=[endpoint_xyz[2]],
            mode="markers+text",
            marker=dict(size=8, color="white", line=dict(color="black", width=2)),
            text=["true endpoint"],
            textposition="middle right",
            name="true endpoint",
            showlegend=False,
            hovertemplate=(
                "true endpoint"
                + f"<br>p(e | traj)={item['true_probability']:.3e}"
                + f"<br>rank={item['true_rank']}<extra></extra>"
            ),
        ),
        row=row,
        col=col,
    )

scene_layout = dict(
    xaxis=dict(visible=False),
    yaxis=dict(visible=False),
    zaxis=dict(visible=False),
    aspectmode="data",
)
fig.update_layout(
    title=dict(text="Example trajectory posterior maps: p(e | traj)", y=0.985),
    height=960,
    margin=dict(l=0, r=0, t=115, b=0),
)
for annotation in fig.layout.annotations:
    annotation.y -= 0.018
for scene_name in [f"scene{i}" if i > 1 else "scene" for i in range(1, 7)]:
    fig.layout[scene_name].update(scene_layout)
fig.show()
````
